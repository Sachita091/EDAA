```python
import numpy as np
arr = np.arange(1,21)

print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
```

```python
np.random.seed(1)
arr = np.random.randint(50,size=(4,5)).reshape(4,5)
print(arr)
print(arr.itemsize)
print(arr.nbytes)
```

```python
np.random.seed(1)
a = np.random.randint(10,size=(3,4))
print(a)
print(a[:2,:2])
print(a[::2])
print(a[:,::-1])
```
