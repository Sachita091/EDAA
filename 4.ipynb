```python
import numpy as np
import pandas as pd

data = pd.Series([1,2,3,4,5])
print(data)

data = pd.Series([1,2,3,4,5],index=[20,30,50,10,90])
print(data)

data1 = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(data1)
```
