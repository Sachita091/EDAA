```python
import pandas as pd
import numpy as np
#using tuples
index = [('a',2010),('a',2020),('a',2025),('b',2010),('b',2020),('b',2025)]
population = [123,456,789,101,123,121]
i = pd.MultiIndex.from_tuples(index,names=["Country","Year"])
print(i)
```

```python
#Series
pop = pd.Series(population,index=i)
print(pop)
#Stack and Unstack
pop_df = pop.unstack()
print(pop_df)
pop_df.stack()
```

```python
#using arrays
# repeating elements to ensure both arrays have length 6
i1 = pd.MultiIndex.from_arrays(
    [['A', 'A', 'A', 'B', 'B', 'B'], [2010, 2020, 2025, 2010, 2020, 2025]],
    names=["Country", "Year"]
)

pop1 = pd.Series(population, index=i1)
print(pop1)
```

```python
#using products
i2 = pd.MultiIndex.from_product([['A','B'],[2010,2020,2025]],names=["Country","Year"])
pop2 = pd.Series(population,index=i2)
print(pop2)
```
